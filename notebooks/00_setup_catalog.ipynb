```python
%sql
CREATE CATALOG transit
  COMMENT 'MBTA transit lakehouse — GTFS static and GTFS-Realtime';
```

```python
%sql
USE CATALOG transit;

CREATE SCHEMA IF NOT EXISTS bronze
  COMMENT 'Raw as-landed. All columns string. Ingestion metadata only, no business logic.';

CREATE SCHEMA IF NOT EXISTS silver
  COMMENT 'Typed, deduplicated, conformed. Business rules and quality gates applied.';

CREATE SCHEMA IF NOT EXISTS gold
  COMMENT 'Star schema. Facts and conformed dimensions for consumption.';

CREATE SCHEMA IF NOT EXISTS ops
  COMMENT 'Pipeline operations: ingestion log, watermarks, quality results, rejects.';

CREATE VOLUME IF NOT EXISTS transit.bronze.landing
  COMMENT 'Landing zone for raw source files before they become Delta tables.';
```

```python
base = "/Volumes/transit/bronze/landing"
for p in ["static", "rt/vehicle_positions"]:
    dbutils.fs.mkdirs(f"{base}/{p}")

display(dbutils.fs.ls(base))
```
